```python
c1.columns=c1.columns.str.lower()                                          # Making all the columns in sneak cases means lower case with underscore.
c1.columns=c1.columns.str.replace(" ","_")                                                
c1=c1.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
c1.columns
```

```python
# create a new column
labels=['Young Adults','Adults','Middle-aged','Seniors']                     
c1['age_group']=pd.qcut(c1['age'],q=4,labels=labels)           # qcut is used to divide the age into 4 parts and with given labels.
```

```python
c1[['age','age_group']]
```

```python
# Create new column purchase_frequency_days

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

c1['purchase_frequency_days']=c1['frequency_of_purchases'].map(frequency_mapping)
```
